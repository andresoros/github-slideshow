
 # Usando el atributo .cdf Poisson calcula las probabilidades anteriores.
* ¿Cuál es la probabilidad de que en un día ocurran más de dos accidentes?


```python
from scipy.stats import poisson

# Calcular P(X <= 2)
probabilidadMenor2 = poisson.cdf(2, 2)

# Calcular P(X > 2) = 1 - P(X <= 2)
probabilidadMayor2 = 1 - probabilidadMenor2

print(probabilidadMayor2)
```

* ¿Cuál es la probabilidad de que en un día ocurran más de dos accidentes, sabiendo que ocurre por lo menos uno?

```python
from scipy.stats import poisson

# Calcular unicamente P(X>2) ya que P(X=1) ya esta contenida en P(X>2)
probabilidadMenor2 = poisson.cdf(2, 2)

# Calcular P(X > 2) = 1 - P(X <= 2)
probabilidadMayor2 = 1 - probabilidadMenor2

#Obtenemos la probabilidad de obtener 0 accidentes
probabilidadMayorUnaccidente = poisson.cdf(0,2)
probabilidadMenosUno = 1- probabilidadMayorUnaccidente

ProbabilidadCondicional = probabilidadMayor2/probabilidadMenosUno
ProbabilidadCondicional
```

Un examen de Estadística consta de 20 preguntas tipo test y se conoce de experiencias anteriores que un alumno tiene probabilidad 0.7 de contestar bien cada pregunta. Obtener:

a) La probabilidad de que la primera pregunta que contesta bien sea la cuarta.

P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}

b) Sabiendo que para aprobar el examen es necesario contestar bien a 10 preguntas, ¿cuál es la probabilidad de que apruebe al contestar la pregunta duodécima?

```python
#Se obtiene la probabilidad de que las 3 primeras preguntas sean incorrectas

# Probabilidades
unaCorrecta = 0.7
unaIncorrecta = 1- unaCorrecta

tresIncorrectas = (unaIncorrecta ** 3)


cuartaCorrecta = tresIncorrectas * unaIncorrecta

cuartaCorrecta
```

b) Sabiendo que para aprobar el examen es necesario contestar bien a 10 preguntas, ¿cuál es la probabilidad de que apruebe al contestar la pregunta duodécima?



```python
from scipy.stats import binom


n = 11  # número de preguntas consideradas
p = 0.7  # probabilidad de contestar bien

# Caso 1: Pregunta 12 correcta (necesita al menos 9 correctas de las 11)
# Se suman las probabiidades de que se tengan 9, 10 y 11 respuestas correctas
# (esto por el numero de preguntas que se llevan hasta llegar a la preguta 12)

prob9Correctas = sum(binom.pmf(k, n, p) for k in range(9, 12))

# Caso 2: Pregunta 12 incorrecta (necesita al menos 10 correctas de las 11)
prob10Correctas = sum(binom.pmf(k, n, p) for k in range(10, 12))

#Se suman ambas probabilidades siguiendo la distribucion binomial
sumaProbabilidades = prob9Correctas * p + prob10Correctas * (1 - p)

print(f"Probabilidad de aprobar al contestar la pregunta duodécima: {sumaProbabilidades:.4f}")
```

La cantidad que pones de nuevo forma una serie infinita:

1+ 0.5 + 0.25 + 0.125 + …

Esta es una serie geométrica, donde cada término se obtiene multiplicando el anterior por una constante.

Fórmula de la Suma de una Serie Geométrica
La suma de una serie geométrica infinita se puede calcular usando la fórmula:

$$ S = \frac{a}{1 - r} $$


donde:


S es la suma total de la serie.

a es el primer término de la serie.

r es la razón común en este caso
 $$ r=  \frac{1}{2} $$


Aplicando la Fórmula
Identificamos
𝑎 y 𝑟

El primer término
𝑎 = 1

La razón $$r= \frac{1}{2}$$

Sustituyendo en la fórmula:

$$S = \frac{1}{1 - \frac{1}{2}} = 2$$

Una compañía petrolera realiza un estudio geológico que indica que un pozo petrolero exploratorio debería tener un 20% de posibilidades de encontrar petróleo.

¿Cuál es la probabilidad de que el primer pozo se produzca en el tercer pozo perforado?

```python

p = 0.2  # probabilidad de encontrar petróleo
q = 1 - p  # probabilidad de no encontrar petróleo

k = 3  # tercer pozo

#Usando distribución geométrica.
probabilidad = (q ** (k - 1)) * p

print(f"La probabilidad de que el primer pozo en encontrar petróleo sea el tercer pozo es: {probabilidad:.3f}")
```

¿Cuál es la probabilidad de que el tercer pozo se produzca en el séptimo pozo perforado?

```python
from math import comb

# Probabilidades
p = 0.2  # probabilidad de encontrar petróleo
q = 1 - p  # probabilidad de no encontrar petróleo

# Parámetros
k = 3  # número de éxitos deseados
n = 7  # total de ensayos

# Cálculo de la probabilidad
coef_binom = comb(n - 1, k - 1)  # Coeficiente binomial
probabilidad = coef_binom * (p ** k) * (q ** (n - k))

print(f"La probabilidad de que el tercer pozo en encontrar petróleo sea el séptimo pozo perforado es: {probabilidad:.4f}")
```

¿Cuál es la media y la varianza del número de pozos que se deben perforar si la compañía petrolera quiere establecer tres pozos productores?

```python
# Probabilidades y parámetros
p = 0.2  # probabilidad de encontrar petróleo
k = 3    # número de pozos productores deseados

# Cálculo de la media
media = k / p

# Cálculo de la varianza
varianza = (k * (1 - p)) / (p ** 2)

# Resultados
print(f"La media del número de pozos que se deben perforar es: {media:.2f}")
print(f"La varianza del número de pozos que se deben perforar es: {varianza:.2f}")
```
